```python
class Student:
    def __init__(self,name,age,roll,marks):
        self.full_name=name
        self.age=age
        self.roll=roll
        self.marks=marks
# create object
S1=Student("Vishnu",24,14,359)
S2=Student("Rahul",25,1,400)
# name=Vishnu
# age=24
# roll=14
# marks=359
print(S1.full_name)
print(S1.roll)
print(S1.marks)
print(S2.full_name)
```

```python
# Method
Class ke ander bnaya gya function ko hum method kehte hai.
```

```python
class Student:#class
    def __init__(self,name,age,marks):
        self.name=name# attributes
        self.age=age
        self.marks=marks
    def display(self):# method
        print(self.name)
        print(self.marks)
S1=Student("Rahul",24,400)
#s1  object
S1.display()  
```
