```python
class Student:
    def __init__(self,name,age,roll,marks):
        self.full_name=name
        self.age=age
        self.roll=roll
        self.marks=marks
# create object
S1=Student("Vishnu",24,14,359)
S2=Student("Rahul",25,1,400)
# name=Vishnu
# age=24
# roll=14
# marks=359
print(S1.full_name)
print(S1.roll)
print(S1.marks)
print(S2.full_name)
```

```python
class BankAccount:
    def __init__(self,name,balance):
        self.name=name
        self.__balance=balance
    def deposit(self,amount):
        if amount>0:
            self.__balance+=amount
            print("Amount Deposited Successfully")
        else:
            print("Invalid Amount Entered")
    def withdraw(self,amount):
        if amount<=0:
            print("Invalid Amount")
        elif amount>self.__balance:
            print("Insufficient Amount")
        else:
            self.__balance-=amount
            print("Withdraw Successfully")
    def get_balance(self):
        return self.__balance

account=BankAccount("Rahul",100000)

```

```python
class BankAccount:
    def __init__(self,name,balance):
        self.name=name
        self.__balance=balance
    def deposit(self,amount):
        if amount>0:
            self.__balance+=amount
            print("Amount Deposited Successfully")
        else:
            print("Invalid Amount Entered")
    def withdraw(self,amount):
        if amount>50000:
            print("Daily Withdrawal Limit Exceeded")
        elif amount>self.__balance:
            print("Insufficient Amount")
        else:
            self.__balance-=amount
            print("Withdraw Successfully")
    def get_balance(self):
        return self.__balance

account=BankAccount("Rahul",100000)
```
