# Dictionary Datatype    Class Dict


1. Dictionary is a collection Datatype

2. Dictionary  store data in key- value Pairs

3. Dictionary doesn't Support Indexing or SLicing

4. Dictionary is a Mutable/Changeble  Datatype

5. Dictionary Never Store Duplicates Key

6. Dictionary Literal is {} Curlie Braket Sprated by COLON (:)

7. Example 
        a={1:"Vishn Ghosh"}

        1-----Key

        Vishnu Ghosh  ----Value

```python
# Create Dictionary
records={1:"Ram",2:"Shyam",3:"Raghav",4:"Koyal"}
print(records)
```

```python
print(type(records))
```

```python
# Adding Element In Existing Dict
records[5]="Rimpa"
print(records)
```

```python
# Update Element Value

records[3]="Jyoti"
print(records)
```

```python
records[5]="Rimpa"
print(records)
```

```python
a={1: 'Ram', 2: 'Shyam', 3: 'Jyoti', 4: 'Koyal', 5: 'Rimpa',5:"Vishnu"}  # Never Support Duplicates
print(a)
```

```python
print(a.items())
```

```python
print(a.keys())
```

```python
print(a.values())
```
