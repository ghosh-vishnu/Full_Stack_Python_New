```python
def check_number(number):
    if number %2 ==0:
        return "Even Number"
    else:
        return "Odd Number"
print(check_number(85))
print(check_number(100))
```

```python
# Function+Loop

def series(number):
    for i in range(1,number+1):
        print(i)
print(series(10))
```

```python
# Function+Loop

def series(number):
    numbers=[]
    for i in range(1,number+1):
        numbers.append(i)
    # print(numbers)
    for i in numbers:
        print(i)
    return numbers
    # for i in numbers:
    #     print(i)

result=series(10)
# for i in result:
#     print(i)
# # print(result)
```

```python
def table_square(n):
    if n >20:
        return "No Find Out"

    for i in range(1,11):
        result=n*i
        square=result*result
        print(f"{n} X {i}= {result} square = {square}")
number_input=int(input("Enter Table Number"))
table_square(number_input)
```
