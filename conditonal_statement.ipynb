```python
# Comparison Operator

==       Equal                a==b
!=      not Equal             a!=b
>      Greater than           a>b
<      Less Than              a<b
>=    Greather than Equal     a>=b
<=    Less than Equal         a<=b
```

```python
user_pin=int(input("Enter Your PIN"))
PIN=1234
balance=5000
if user_pin==PIN:
    print("Login Successfully")
    print("1. Show Balance")
    print("2. Withdraw Balance")
    print("3. Deposit Balance")
    print("4. Transfer Balance")
    choice=int(input("Please Select Choice"))
    if choice==1:
        print(f"Your Balance is {balance} ")
    elif choice==2:
        withdraw_amount=int(input("Enter Your Withdrawal Amount"))
        if withdraw_amount<=balance:
            if withdraw_amount>=0:
                print(f"Withdraw Succesfully! Current Balance is {balance-withdraw_amount}")
            else:
                print("Please Enter Valid Amount")
        else:
            print("Insufficient Balance in your Account")
    elif choice==3:
        print("Implement Deposit Logic")
    elif choice==4:
        print("Implement Transfer Balance Logic")
    else:
        print("Wrong Choice Entered")
else:
    print("Wrong PIN Entered")
```
