# Match Case Statement

```python
day=10
if day==2:
    print("Monday")
elif day==3:
    print("Tuesday")
# else:
#     print("Invalid Day Number Entered")
```

```python
day=5
match day:
    case 1:
        print("Sunday")
    case 2:
        print("Monday")
    case 3:
        print("Tuesday")
```

```python
Python;s Match case is useful when you need to compare one value against many possible values.
it was introduced in Python 3.10
```

```python
# Default Case _
# use _ when none of the case match.

day=2
match day:
    case 1:
        print("Sunday")
    case 2:
        print("Monday")
    case _:
        print("Invalid Day")
```

```python
# Multiple Values in One Case
day=input("Enter Day Name").lower()
print(f"Your Entered Day Name is {day}")
match day:
    case "saturday" | "sunday":
        print("Weekend Days")
    case 'monday' |'tuesday'|'wednesday'| "thursday"|'friday':
        print("Working Days")
    case _:
        print("Invalid Day Name Entered")
```

```python
# Math-case with sting
command='start'
match command:
    case 'start':
        print("Starting Application")
    case 'stop':
        print("Stoping Application")
    case 'restart':
        print("Application Restarting")
    case _:
        print("Unknown Command")
    
```

```python
# Math-case with conditions -Guard
age=61
match age:
    case x if x<18:
        print("Minor")
    case x if x >=18 and x<60:
        print("Adult")
    case x if x>60:
        print("Senior Citizen")
```

```python
# Real World Example -User Role
role='Admin'.lower()
match role:
    case 'admin':
        print("Full Access")
    case 'manage':
        print("Manager Access")
    case 'employee':
        print("Employee Access")
    case 'guest':
        print("Guest Access")
    case _:
        print('Your Role is Not Defined')
```
