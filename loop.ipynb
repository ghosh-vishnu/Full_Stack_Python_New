``````python
n=5
for i in range(n):
    for j in range(n):
        if i==0 or i==n-1 or j==0 or j==n-1:
            print("*",end=" ")
        else:
            print(" ",end=" ")
    print()
``````

``````python
n=5
for i in range(1,n+1):
    for j in range(n-i):
        print(" ",end=" ")
    for j in range(2*i-1):
        print("*",end=" ")
    print()
``````

``````python
n = 5
for i in range(n):
    for j in range(n):
        if i == j or i + j == n - 1:
            print(j, end=" ")
        else:
            print(" ", end=" ")

    print()
``````

``````python
n = 5
for i in range(n):
    # Spaces
    for j in range(i if i <= 2 else n - 1 - i):
        print("  ", end="")
    # Stars
    for j in range(5 - 2 * (i if i <= 2 else n - 1 - i)):
        print("* ", end="")
    print()
``````
