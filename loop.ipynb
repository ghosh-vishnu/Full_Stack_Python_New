```python
number=[10,20,30,50,60,70,80,55,64,15,23,87,90]
odd=[]
even=[]
odd_sum=0
even_sum=0
for i in number:
    if i%2==0:
        even.append(i)
        even_sum+=i
    else:
        odd.append(i)
        odd_sum+=i
print(f"Odd List is {odd} and the sum of these all odd number is {odd_sum}")
print(f"Even List is {even} and the sum if these all even number is {even_sum}")

```

# While Loop

```python
While Loop--A while loop is used when you need to repeat a task as long as a contiditon is True.
```

```python
count=1
while count <=5:
    print(count)
    count+=1
```

```python
count =5
while count >=1:
    print(count)
    count-=1
```

```python
password=""
count=0
user_password='1234'
while password!=user_password:
    password=input("Enter Your Password")
    count+=1
print("Login Successfully")
print(f"Total Attempts",count)
```

# break

```python
count=1
while count<=10:
    print(count)
    if count==5:
        break
    count+=1
```

# continue   
Use for skip continue itteration 

```python
count=0
while count<5:
    count+=1
    if count==3:
        continue
    print(count)
```

```python
for i in range(1,6):
    if i==3:
        break
    print(i)
```

```python
for i in range(1,6):
    if i==3:
        continue
    print(i)
```

# Nested Loop

Loop inside Loop is called Nested Loop

```python
for i in range(3):
    for j in range(3):
        print(i,j)
```

```python
for i in range(1,5):
    sum=0
    for j in range(1,11):
        print(i,"X",j,"=",i*j)
        sum+=i*j
    print(sum)
    print()
```

```python
for i in range(5):
    for j in range(5):
        print('*',end=" ")
    print()
```

```python
students=[
    ['Vishnu',25],
    ['Rahul',22],
    ['Amit',24],
    ['Rimpa',20]
]
for student in students:
    for data in student:
        print(data)
    print()

```

# Nested Loop+ Condition

```python
for i in range(1,6):
    for j in range(1,6):
        if j==3:
            print("*",end=" ")
        else:
            print(j,end=" ")
    print()
```
