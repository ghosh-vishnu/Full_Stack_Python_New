```python
# Loop through String
name="Vishnu"
# V
# I
# S
# H
# N
# U
for char in name:
    print(char.upper())
```

```python
# Loop Through a List
data=['Apple','Banana','Mango']
for fruit in data:
    print(fruit.upper(),len(fruit))

# APPLE 5
# BANANA 6
# MANGO 5
```

```python
# Loop through Tuple
number=(10,20,30,40,50,60)
total=0
for i in number:
    print(i)
    total+=i
print(f"The Sum of Tuple is {total}")
```

```python
# Loop Through Dictionary
student={
    'name':"Vishnu",
    'age':25,
    'Class':'MCA'
}
for key,value in student.items():
    print(key,value)
```

```python
number=[10,20,30,50,60,70,80,55,64,15,23,87,90]
odd=[]
even=[]
odd_sum=0
even_sum=0
for i in number:
    if i%2==0:
        even.append(i)
        even_sum+=i
    else:
        odd.append(i)
        odd_sum+=i
print(f"Odd List is {odd} and the sum of these all odd number is {odd_sum}")
print(f"Even List is {even} and the sum if these all even number is {even_sum}")

```
